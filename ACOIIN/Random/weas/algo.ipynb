# Taller de Programación Competitiva: Complejidad Algorítmica (Big-O)
**Bienvenido al taller.** En programación competitiva (OCI, ICPC, Codeforces), no basta con que un algoritmo resuelva el problema; debe hacerlo **dentro del límite de tiempo** (usualmente 1 o 2 segundos).

---

## 1. ¿Por qué no usamos un cronómetro?

Si medimos el tiempo de ejecución en segundos con un reloj, los resultados varían según:
* La potencia del procesador de cada computador.
* Si tienes 40 pestañas de Chrome abiertas o Discord de fondo.
* El lenguaje de programación y las optimizaciones del compilador.

Para comparar algoritmos de forma justa e independiente del hardware, medimos **cómo crece la cantidad de operaciones elementales a medida que el tamaño de la entrada ($N$) aumenta**. A esto le llamamos **Notación Asintótica o Big-O** ($\mathcal{O}$).

## 2. La Regla de Oro de la Programación Competitiva ⏱️

> **En un juez en línea estándar, un computador moderno ejecuta aproximadamente $10^8$ (100 millones) de operaciones básicas por segundo.**

Mirando las restricciones de $N$ en el enunciado del problema, puedes deducir de inmediato qué complejidad debes diseñar:

| Valor máximo de $N$ | Complejidad esperada | Ejemplo típico |
| :--- | :--- | :--- |
| $N \le 10..12$ | $\mathcal{O}(N!)$ o $\mathcal{O}(2^N \cdot N)$ | Permutaciones, Backtracking exhaustivo |
| $N \le 20..25$ | $\mathcal{O}(2^N)$ | Generación de subconjuntos, Bitmask DP |
| $N \le 500$ | $\mathcal{O}(N^3)$ | Tres bucles anidados, Floyd-Warshall |
| $N \le 5.000$ | $\mathcal{O}(N^2)$ | Dos bucles anidados, ordenamiento básico |
| $N \le 2 \cdot 10^5$ | $\mathcal{O}(N \log N)$ | Ordenamiento rápido, Segment Tree, Dijkstra |
| $N \le 10^7$ | $\mathcal{O}(N)$ | Un solo bucle lineal, Two Pointers, Criba |
| $N \ge 10^9$ | $\mathcal{O}(\log N)$ o $\mathcal{O}(1)$ | Búsqueda Binaria, Fórmulas matemáticas |

## 3. Visualizando las Curvas de Complejidad
Ejecuta la siguiente celda para ver gráficamente cómo escalan las operaciones a medida que $N$ crece.

````python
import numpy as np
import matplotlib.pyplot as plt

# Rango de prueba pequeño para que las exponenciales no exploten
n = np.linspace(1, 20, 100)

plt.figure(figsize=(9, 6))
plt.plot(n, np.ones_like(n), label="O(1) - Constante", linewidth=2)
plt.plot(n, np.log2(n), label="O(log N) - Logarítmica", linewidth=2)
plt.plot(n, n, label="O(N) - Lineal", linewidth=2)
plt.plot(n, n * np.log2(n), label="O(N log N) - Casi lineal", linewidth=2)
plt.plot(n, n**2, label="O(N^2) - Cuadrática", linewidth=2)
plt.plot(n, 2**n, label="O(2^N) - Exponencial (¡Peligro!)", linewidth=2, linestyle="--")

plt.ylim(0, 100)
plt.xlim(1, 20)
plt.xlabel("Tamaño de la entrada (N)", fontsize=12)
plt.ylabel("Operaciones aproximadas", fontsize=12)
plt.title("Comparativa de Complejidades Temporales", fontsize=14)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

## 4. Caso Práctico: ¿Dos números suman K? (Two-Sum)
**Problema:** Dado un arreglo de $N$ números ordenados, queremos saber si existen dos números cuya suma sea exactamente igual a $K$.

### Estrategia A: Fuerza Bruta $\mathcal{O}(N^2)$
Compara cada par posible usando dos bucles anidados.

````python
def suma_dos_fuerza_bruta(arreglo, k):
    n = len(arreglo)
    for i in range(n):
        for j in range(i + 1, n):
            if arreglo[i] + arreglo[j] == k:
                return True
    return False
````

### Estrategia B: Dos Punteros (Two Pointers) $\mathcal{O}(N)$
Aprovecha que el arreglo está ordenado. Un puntero al inicio y otro al final; ajustamos según falte o sobre suma.

````python
def suma_dos_optimo(arreglo, k):
    izq = 0
    der = len(arreglo) - 1
    while izq < der:
        actual = arreglo[izq] + arreglo[der]
        if actual == k:
            return True
        elif actual < k:
            izq += 1
        else:
            der -= 1
    return False
````

### Comparación de rendimiento en vivo
Vamos a generar un arreglo con $N = 10.000$ números y medir el tiempo que tarda cada enfoque.

````python
import time

N = 10000
datos = list(range(N))
objetivo = -1  # Un valor imposible para forzar al algoritmo al peor caso posible

# Medición O(N^2)
t_inicio = time.time()
suma_dos_fuerza_bruta(datos, objetivo)
t_fuerza_bruta = time.time() - t_inicio

# Medición O(N)
t_inicio = time.time()
suma_dos_optimo(datos, objetivo)
t_optimo = time.time() - t_inicio

print(f"Resultados para N = {N}:")
print(f"-> Fuerza Bruta O(N^2): {t_fuerza_bruta:.4f} segundos")
print(f"-> Two Pointers O(N):   {t_optimo:.6f} segundos")
print(f"-> La solución óptima fue aproximadamente {int(t_fuerza_bruta / max(t_optimo, 1e-9))} veces más rápida.")
````
